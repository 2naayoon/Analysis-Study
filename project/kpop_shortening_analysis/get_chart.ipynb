## 유튜브 뮤직 국내 차트

### 유튜브 뮤직 차트 csv 다운로드

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
from datetime import datetime, timedelta
import undetected_chromedriver as uc

options = uc.ChromeOptions()
options.add_experimental_option("prefs", {"download.default_directory": "./"})
options.add_argument("--disable-blink-features=AutomationControlled")

driver = uc.Chrome(options=options, use_subprocess=False)
# 아래 당신의 기존 코드 로직을 그대로 유지하세요.
# 셀렉터 모두 원본 코드 그대로 작동합니다.

dates = []

# (오늘일자-5일)로부터 3년간 weekly 날짜 가져오기
current_date = datetime.now() - timedelta(days=5)
end_date = current_date - timedelta(days=(2 * 365))
while current_date >= end_date:
    date_string = current_date.strftime("%Y%m%d")
    dates.append(date_string)
    current_date -= timedelta(weeks=1)

with webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options) as driver:
    driver.get("https://charts.youtube.com/charts/TopSongs/kr/weekly")
    
    time.sleep(5)
    
    # CSV 파일 다운로드
    for date in dates:
        driver.get(f"https://charts.youtube.com/charts/TopSongs/kr/weekly/{date}")

        time.sleep(3)

        csv_download_button = driver.find_element(By.XPATH, '//*[@id="download-button"]')
        csv_download_button.click()
```

### 국내 차트 통합

```python
import pandas as pd
import os

# 통합 csv dataframe 생성
header = ['Rank', 'Previous Rank', 'Track Name', 'Artist Names', 'Periods on Chart', 'Views', 'YouTube URL']
combined_df = pd.DataFrame(columns=header)

# 파일명을 기반으로 국가명과 일자 정보를 추출하여 데이터에 추가
directory = 'kr_top_chart_csv_2023-2025/'
for filename in os.listdir(directory):
    if filename.endswith('.csv'):
        file_path = os.path.join(directory, filename)
        
        # 확장자 제거 후 파일명 분할
        name_parts = filename.replace('.csv', '').split('-')
        
        try:
            country = name_parts[4]  # 'kr'
            date_str = name_parts[-1]  # '20230713'
            date_formatted = f"{date_str[:4]}-{date_str[4:6]}-{date_str[6:]}"
            
            print(f"처리 중: {filename} -> 국가: {country}, 날짜: {date_formatted}")
            
            # 기존 데이터 로드
            df = pd.read_csv(file_path)
            
            # 국가와 날짜 추가
            df['Country'] = country
            df['Date'] = date_formatted
            
            # DataFrame을 통합
            combined_df = pd.concat([combined_df, df], ignore_index=True)
        
        except IndexError:
            print(f"파일명 파싱 실패: {filename}")
            continue

# 저장 디렉토리 확인 및 저장
os.makedirs('data', exist_ok=True)
new_file_path = os.path.join('data', 'kr_top_combined.csv')
combined_df.to_csv(new_file_path, index=False)
print(f"\n✅ 저장 완료: {new_file_path}")
```

## Spotify 글로벌 차트

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import time
from datetime import datetime, timedelta

options = webdriver.ChromeOptions() ;
prefs = {"download.default_directory" : "global_top_chart_csv_2023-2025"};
options.add_experimental_option("prefs",prefs);

dates = []

# (오늘일자-5일)로부터 3년간 weekly 날짜 가져오기
current_date = datetime.now() - timedelta(days=5)
end_date = current_date - timedelta(days=(3 * 365))
while current_date >= end_date:
    date_string = current_date.strftime("%Y-%m-%d")
    dates.append(date_string)
    current_date -= timedelta(weeks=1)

with webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options) as driver:
    driver.get("https://charts.spotify.com/charts/view/regional-global-weekly/latest")
    time.sleep(2)
    
    # 로그인 필요함
    login_button = driver.find_element(By.XPATH, '//*[@id="__next"]/div/div/header/div/div[2]/a/span[1]')
    login_button.click()

    id_input = driver.find_element(By.XPATH, '//*[@id="login-username"]').send_keys('user-id')

    cont_button = driver.find_element(By.XPATH, '//*[@id="login-button"]')
    cont_button.click()

    time.sleep(2)

    password_button = driver.find_element(By.XPATH, '//*[@id="encore-web-main-content"]/div[2]/div/div/div/form/div[2]/section/button')
    password_button.click()

    time.sleep(2)
    
    # 로그인 화면 진입
    password_input = driver.find_element(By.XPATH, '//*[@id="login-password"]').send_keys('password')
    login_button = driver.find_element(By.XPATH, '//*[@id="login-button"]')
    login_button.click()
    
    time.sleep(5)
    
    # CSV 파일 다운로드
    for date in dates:
        driver.get(f"https://charts.spotify.com/charts/view/regional-global-weekly/{date}")

        time.sleep(3)

        csv_download_button = driver.find_element(By.XPATH, '//*[@id="__next"]/div/div[3]/div/div/div[2]/span/span/button')
        csv_download_button.click()
```

### Spotify 음원 길이 추가 후 파일 통합

```python
import pandas as pd
import os

# 통합 csv dataframe 생성
header = ['rank', 'uri', 'artist_name', 'track_name', 'source', 'peak_rank', 'previous_rank', 'weeks_on_chart', 'streams', 'duration_sec']
combined_df = pd.DataFrame(columns=header)

# 파일이 들어 있는 디렉토리
directory = 'global_duration/'

# 파일명을 기반으로 국가명과 날짜 정보 추출
for filename in os.listdir(directory):
    if filename.endswith('.csv'):
        file_path = os.path.join(directory, filename)

        # 파일명 예시: regional-global-weekly-2023-01-05.csv
        name_parts = filename.replace('.csv', '').split('-')

        try:
            country = name_parts[1]  # ex: 'global'
            date_str = name_parts[-3] + '-' + name_parts[-2] + '-' + name_parts[-1]  # '2023-01-05'
            
            print(f"처리 중: {filename} -> 국가: {country}, 날짜: {date_str}")
            
            # 기존 데이터 로드
            df = pd.read_csv(file_path)

            # 날짜 및 국가 컬럼 추가
            df['Date'] = date_str
            df['Country'] = country

            # DataFrame 통합
            combined_df = pd.concat([combined_df, df], ignore_index=True)

        except IndexError:
            print(f"❌ 파일명 파싱 실패: {filename}")
            continue

# 저장 디렉토리 확인 및 저장
os.makedirs('data', exist_ok=True)
new_file_path = os.path.join('data', 'global_top_combined.csv')
combined_df.to_csv(new_file_path, index=False)
print(f"\n✅ 저장 완료: {new_file_path}")
```
